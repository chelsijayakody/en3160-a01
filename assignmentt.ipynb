```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt
```

```python
image= cv.imread("images/q1.jpeg", cv.IMREAD_GRAYSCALE)
```

```python
print(image.shape)
print(image.dtype)
```

```python
plt.imshow(image, cmap="gray")
plt.title("Original Image")
plt.axis("off")
plt.show()
```

```python
breakpoints = np.array([
    [0,   0],
    [50,  50],
    [100, 150],
    [150, 255],
    [255, 255]
])
```

```python
def intensity_transform(im, breakpoints):
    x = breakpoints[:, 0]
    y = breakpoints[:, 1]

    output = np.interp(im, x, y)

    return output.astype(np.uint8)
```

```python
transformed = intensity_transform(
    image,
    breakpoints
)
```

```python
plt.imshow(transformed, cmap="gray")
plt.title("Transformed Image")
plt.axis("off")
plt.show()
```

```python
plt.figure(figsize=(6, 6))

plt.plot(
    breakpoints[:, 0],
    breakpoints[:, 1],
    marker="o"
)

plt.xlabel("Input intensity")
plt.ylabel("Output intensity")
plt.title("Intensity Transformation")

plt.xlim(0, 255)
plt.ylim(0, 255)

plt.grid()
plt.show()
```

```python
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(image, cmap="gray")
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(transformed, cmap="gray")
plt.title("Transformed Image")
plt.axis("off")

plt.tight_layout()
plt.show()
```
